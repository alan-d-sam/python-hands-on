- Create a class for students with 3 attribute name, class and roll number and assign them using constructor
- add a method to display all the values
- make the class variable protected and roll number private
- create a getter and setter for class name and roll number

```python
class student:
  def __init__(self,name,class_name,roll_number):
    self.name=name
    self._class_name=class_name
    self.__roll_number=roll_number

  def display(self):
    print(f"""Name : {self.name}
          \nClass : {self._class_name}
          \nRoll No : {self.__roll_number}""")

  def getclass_name(self):
    print(self._class_name)

  def setclass_name(self,class_name):
    self._class_name=class_name
    print(self._class_name)

  def setroll_number(self,roll_number):
    self.__roll_number=roll_number
    print(self.__roll_number)

  def getroll_number(self):
    print(self.__roll_number)
```

- Write a program to implement inheritance where an animal class will have
- attributes such as name, _age
- methods will be such as speak and getter and setter for age
- then, there will be 3 child classes lion,dogs, and cats with the same features as animal class but with extra method that will be animal type


```python
class animal():
  def __init__(self,name,age):
    self.name=name
    self._age=age

  def speak(self):
    print("Animal speaks")

  def getage(self):
    print(self._age)

  def setage(self,age):
    self._age=age
    print(self._age)

class lion(animal):
  def __init__(self,name,age,breed):
    super().__init__(name,age)
    self.breed=breed

  def animaltype(self):
    print("Lion")

  def display(self):
    print(f"""Name : {self.name}
          \nAge : {self._age}
          \nBreed : {self.breed}""")

class dogs(animal):
  def __init__(self,name,age,breed):
    super().__init__(name,age)
    self.breed=breed

  def animaltype(self):
    print("Dogs")

  def display(self):
    print(f"""Name : {self.name}
          \nAge : {self._age}
          \nBreed : {self.breed}""")

class cat(animal):
  def __init__(self,name,age,breed):
    super().__init__(name,age)
    self.breed=breed

  def animaltype(self):
    print("Cat")

  def display(self):
    print(f"""Name : {self.name}
          \nAge : {self._age}
          \nBreed : {self.breed}""")
```

- implement polymorphism in the above example
- create speak() method in all the child classes and display different results

```python
class animal():
  def __init__(self,name,age):
    self.name=name
    self._age=age

  def speak(self):
    print("Animal speaks")

  def getage(self):
    print(self._age)

  def setage(self,age):
    self._age=age
    print(self._age)

class lion(animal):
  def __init__(self,name,age,breed):
    super().__init__(name,age)
    self.breed=breed

  def animaltype(self):
    print("Lion")

  def speak(self):
    print("Roar")

  def display(self):
    print(f"""Name : {self.name}
          \nAge : {self._age}
          \nBreed : {self.breed}""")

class dogs(animal):
  def __init__(self,name,age,breed):
    super().__init__(name,age)
    self.breed=breed

  def animaltype(self):
    print("Dogs")

  def speak(self):
    print("Bark")

  def display(self):
    print(f"""Name : {self.name}
          \nAge : {self._age}
          \nBreed : {self.breed}""")

class cat(animal):
  def __init__(self,name,age,breed):
    super().__init__(name,age)
    self.breed=breed

  def animaltype(self):
    print("Cat")

  def speak(self):
    print("Meow")

  def display(self):
    print(f"""Name : {self.name}
          \nAge : {self._age}
          \nBreed : {self.breed}""")
```

- Implement duck typing for the above example

```python
def speak(obj):
    obj.speak()

l = lion("Lion name", 5, "lion")
d = dogs("dog name", 3, "dog")
c = cat("cat name", 2, "cat")

speak(l)
speak(d)
speak(c)
```

- Use abstraction to make a class vehicle and their child class car and motorcycle in which accelerate(), brake() methods are mandatory and there can be any other methods


```python
from abc import ABC, abstractmethod

class vehicle(ABC):
  @abstractmethod
  def accelerate(self):
    print("accelerate")

  @abstractmethod
  def brake(self):
    print("brake")

class car(vehicle):
  def accelerate(self):
    print("accelerate")

  def brake(self):
    print("brake")

class motorcycle(vehicle):
  def accelerate(self):
    print("accelerate")

  def brake(self):
    print("brake")
```
